```python
# ライブラリのインポート

import os

from ultralytics import YOLO
```

```python
# ディレクトリの定義

prepared_dir = os.path.join("data", "prepared")
```

```python
# モデルのトレーニング

model_path = os.path.join("weights", "yolo12l.pt")
model = YOLO(model_path)

model.train(
    data=os.path.join(prepared_dir, "data.yaml"),
    epochs=120,          # 100
    patience=0,          # 100
    batch=48,            # 16
    imgsz=864,           # 640
    save_period=1,       # -1
    device=0,
    rect=False,          # False
    project="runs",
    name="exp",
    optimizer="SGD",     # auto
    seed=0,              # 0
    cos_lr=True,         # False
    close_mosaic=0,      # 10
    lr0=0.007,           # 0.01
    lrf=0.01,            # 0.01
    weight_decay=0.0005, # 0.0005
    box=8.0,             # 7.5
    cls=0.5,             # 0.5
    dfl=2.0,             # 1.5
    val=False,           # True
    hsv_h=0.025,         # 0.015
    hsv_s=0.7,           # 0.7
    hsv_v=0.5,           # 0.4
    degrees=2.0,         # 0.0
    translate=0.1,       # 0.1
    scale=0.5,           # 0.5
    shear=0.0,           # 0.0
    perspective=0.0,     # 0.0
    flipud=0.0,          # 0.0
    fliplr=0.5,          # 0.5
    bgr=0.0,             # 0.0
    mosaic=1.0,          # 1.0
    mixup=0.5,           # 0.0
    cutmix=0.5,          # 0.0
)
```
