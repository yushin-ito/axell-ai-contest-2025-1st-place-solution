```python
# ライブラリのインポート

import glob
import os

import pandas as pd
import torch
from diffusers import StableDiffusion3Pipeline, StableDiffusionLatentUpscalePipeline
from PIL import Image
from transformers import CLIPModel, CLIPProcessor
```

```python
# ディレクトリの定義

generated_dir = os.path.join("data", "generated")
os.makedirs(generated_dir, exist_ok=True)
```

```python
# プロンプトの作成

df = pd.read_csv(os.path.join(generated_dir, "prompts", "background.csv"))
prompts = df["prompt"].tolist()

count = len(glob.glob(os.path.join(generated_dir, "backgrounds", "*.jpg")))

prompts = prompts[count:]
```

```python
# モデルの読み込み

pipeline = StableDiffusion3Pipeline.from_pretrained(
    "stabilityai/stable-diffusion-3.5-large", torch_dtype=torch.bfloat16
)
pipeline = pipeline.to("cuda")

upscaler = StableDiffusionLatentUpscalePipeline.from_pretrained(
    "stabilityai/sd-x2-latent-upscaler", torch_dtype=torch.float16
)
upscaler.to("cuda")

model = CLIPModel.from_pretrained("openai/clip-vit-base-patch32")
model = model.to("cuda")
processor = CLIPProcessor.from_pretrained("openai/clip-vit-base-patch32")
```

```python
# 画像の生成

save_dir = os.path.join(generated_dir, "backgrounds")

threshold = 0.5
generator = torch.manual_seed(42)

for i, prompt in enumerate(prompts, start=count + 1):
    count = 0
    while True:
        count += 1

        image = pipeline(
            prompt,
            generator=generator,
        ).images[0]

        image = upscaler(
            prompt=prompt,
            image=image,
            generator=generator,
        ).images[0]

        inputs = processor(text=[prompt], images=image, return_tensors="pt").to("cuda")

        with torch.no_grad():
            outputs = model(**inputs)
            score = outputs.logits_per_image.item()

        if score < threshold and count < 3:
            continue

        aspect = 16 / 9
        height = int(round(image.height / aspect))
        y = (image.height - height) // 2
        image = image.crop((0, y, image.width, y + height))
        image = image.resize((1920, 1080), Image.LANCZOS)

        save_path = os.path.join(save_dir, f"T{i}.jpg")
        image.save(save_path)

        break
```

```python
# プロンプトの作成

df = pd.read_csv(os.path.join(generated_dir, "prompts", "foreground.csv"))
prompts = df["prompt"].tolist()

count = len(glob.glob(os.path.join(generated_dir, "foregrounds", "*.jpg")))

prompts = prompts[count:]
```

```python
# 画像の生成

save_dir = os.path.join(generated_dir, "foregrounds")

threshold = 0.5
generator = torch.manual_seed(42)

for i, prompt in enumerate(prompts, start=count + 1):
    count = 0
    while True:
        count += 1

        image = pipeline(
            prompt,
            generator=generator,
        ).images[0]

        inputs = processor(text=[prompt], images=image, return_tensors="pt").to("cuda")

        with torch.no_grad():
            outputs = model(**inputs)
            score = outputs.logits_per_image.item()

        if score < threshold and count < 3:
            continue

        save_path = os.path.join(save_dir, f"T{i}.jpg")
        image.save(save_path)

        break
```
